## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd


df = pd.read_csv("content_refresh_anonymized.csv")


total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
pct_declining = (declining_pages / total_pages) * 100
avg_impressions = df["impressions_90d"].mean()
avg_word_count = df["word_count"].mean()

print(f"Total pages in starter dataset: {total_pages:,}")
print(f"Pages with a declining trend: {declining_pages:,} ({pct_declining:.1f}%)")
print(f"Average 90-day impressions per page: {avg_impressions:,.0f}")
print(f"Average word count per page: {avg_word_count:,.0f}")
```
